# Prepare train/val and test data folders

```python
%run ./notebook_init.py

import os
import math
import shutil

from glob import glob
from sklearn.model_selection import train_test_split

from core import DATA_FOLDER
```

```python
root_data_folder = os.path.join(DATA_FOLDER, "processed_data")
png_data_folder = os.path.join(DATA_FOLDER, "png_processed_data")

root_train_folder = os.path.join(DATA_FOLDER, "train_data_root")
root_test_folder  = os.path.join(DATA_FOLDER, "test_data_root")
png_train_folder  = os.path.join(DATA_FOLDER, "train_data_png")
png_test_folder   = os.path.join(DATA_FOLDER, "test_data_png")
```

```python
for folder in [root_train_folder, root_test_folder, png_train_folder, png_test_folder]:
    os.makedirs(folder, exist_ok=True)
```

```python
def split_and_copy(data_folder, train_folder, test_folder, ext):
    categories = [d for d in os.listdir(data_folder) if os.path.isdir(os.path.join(data_folder, d))]

    for category in categories:
        category_path = os.path.join(data_folder, category)
        files = glob(os.path.join(category_path, f"*.{ext}"))

        if not files:
            print(f"No {ext} files found in {category_path}")
            continue

        n = len(files)

        if n <= 5:
            # Small set → always 1 test file
            test_files = [files[0]]
            train_files = files[1:]
        else:
            # Normal split: 15% test
            test_size = max(1, math.floor(0.15 * n))  # at least 1 file
            train_files, test_files = train_test_split(files,
                                                       test_size=test_size,
                                                       random_state=42)


        # Create category dirs
        os.makedirs(os.path.join(train_folder, category), exist_ok=True)
        os.makedirs(os.path.join(test_folder, category), exist_ok=True)

        # Copy files
        for f in train_files:
            shutil.copy(f, os.path.join(train_folder, category))
        for f in test_files:
            shutil.copy(f, os.path.join(test_folder, category))

        print(f"{category} ({ext}): {len(train_files)} train, {len(test_files)} test")
```

```python
split_and_copy(root_data_folder, root_train_folder, root_test_folder, "root")

split_and_copy(png_data_folder, png_train_folder, png_test_folder, "png")
```
